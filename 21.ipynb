Using OpenCV - Video

```python
import cv2
import numpy as np 
```

```python
my_video = cv2.VideoCapture('./data/video-1.mp4')
```

```python
cv2.namedWindow('Video Player', cv2.WINDOW_NORMAL)
```

```python
while True:
    ret, frame = my_video.read()
    
    if not ret:
        break

    cv2.imshow('Video Player', frame)

    if(cv2.waitKey(10)==ord('q')):
        break
        
```

```python
my_video.release
cv2.destroyAllWindows()
```
