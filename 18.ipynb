```python
import cv2
```

```python
image_path = '/Users/dandev947366/Desktop/Projects-github/computer-vision/computer-vision/img/01/birds.png'
img = cv2.imread(image_path)
print(img)
```

```python
img.shape
```

```python
def display(_title, _img):
    cv2.imshow(_title, _img)
    cv2.waitKey(0)
    cv2.destroyWindow(_title)
```

```python
cv2.imshow('Window', img)
cv2.waitKey(0)
cv2.destroyWindow()
```

```python
# Seperate colors
b, g, r = cv2.split(img)

display('Window', img)
display('Red', r)
display('Blue', b)
display('Green', g)

```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
display('Window', img)
display('img_2', img_2)
display('gray_img', gray_img)
```

```python
height, width, z = img.shape
new_h = height/2
new_w = height/2

green_color =(0,255,0)
img2 = img.copy()   
for y in range(int(new_h)):
    for x in range(int(new_w)):
        img2[y,x] = green_color
display('img2', img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
