Computer Vision with Python, OpenCV, Pytorch

Using OpenCV

```python
import cv2
import numpy as np
```

```python
def display(title, img):
    cv2.imshow(title, img)
    cv2.waitKey(0)
    cv2.destroyWindow(title)
```

```python
img = np.zeros((400, 400, 3), dtype=np.uint8)
display('Image', img)
```

```python
# draw straight line
cv2.line(img, (50,50), (350, 50), (255, 0, 0), 2)
display('Image', img)
```

```python
# draw cirle
cv2.circle(img, (100, 100), 50, (0, 255, 0), -1)
display('Image', img)
```

```python
# draw rectangle
cv2.rectangle(img, (100,100), (300,300), (255, 0, 0), -1)
display('Image', img)
```
